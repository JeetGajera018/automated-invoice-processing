```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\Jeet\Desktop\automated-invoice-processing\data\processed\invoices_final.csv")
df.head()
```

```python
plt.figure(figsize=(8, 5))
plt.boxplot(df["total_amount"].dropna(), vert=True)
plt.yscale("log")
plt.title("Distribution of Invoice Total Amounts (log scale)")
plt.ylabel("Total Amount (log scale)")
plt.show()
```

```python
colors = df["is_isolation_forest_anomaly"].map({True: "red", False: "steelblue"})

plt.figure(figsize=(8, 5))
plt.scatter(df["item_count"], df["total_amount"], c=colors, alpha=0.6)
plt.xlabel("Item Count")
plt.ylabel("Total Amount")
plt.title("Invoice Amount vs Item Count (red = Isolation Forest anomaly)")
plt.show()
```
